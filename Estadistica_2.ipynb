# Estadística descriptiva

Esta es una ecuación:

$$
X = \sum_{i=1}^{n} X_i
$$


Al estadisticaestudia fenómenos aleatorios que son aqullos fenómenos o experiencias que, incluso en las mismas condiciones, pueden dar lugar a diferentes resultados.

La estadistica se compone por la estadisctica descriptiva, el calculo de probabilidades y al inferencia estadistica.

### Población e individuo

El conjunto de inidividuos en los que se observa el fenómeno aleatorio recibe el nombre de población. Puede pensarse como el colectivo de estudio conformado por unidades de análisis.

Al conjunto de individuos que elegimos de la poblacion para estudiar lo denominamos muestra. Solamente la obeención de una muestra representativa de la población permitirá obtener conclusiones fiables sobre ella. Hanibualmente estas muestras se obtienen por un procedimiento aleatorio, lo que habitualmente garantiza representatividad.

La uestra será la materia prima a utilizar en la elaboración de las inferencias, y solmaente de buena materia prima se puede obtener buenos productos. La aleatoridad es un objetivo que no siempre es posible alcanzar, pero al que idealmente nos hemos de acercar.

### Variable aleatorio y modelo probabilístico

Habitualmente se desea estimar el valor de alguna caraterística de la población en estudio, digamos la estatura media. Esta valor poblacional lo denominaremos parámetro poblacional o simplemente parámetro. A la características de estudio lo denominamos variable aleatoria X.

El investigador desea entonces realizar una inferencia sobre el parámetro de estudio, el cual puede ser:
1.Estimar el valor (Estimación puntual)
2. Estimar un intervalo numérico en el verosímilmente se encuentra el parámetro (estimación por intervalos de confianza)
3. Decidir si puede considerarse razonable un valor u otro para dicho parámetro (contraste de hipotesis)

El investigador selecciona al azar de la poblacioón unos cuantos individuos n que constituyen la muestra, en los que se observará la variable de estuido X.

Se obtendrán así n realizaciones de la variable aleatorio X, que representaremos por $$(X_{1}, \dots , X_n )$$, entendiéndose cada $$ X_{i}, \ i= 1, \dots , n$$ como el valor que toma la variabel en estudio en el individuo seleccionado al azar en el i-ésimo lugar.

> Sólo ses considerará la situación en la que cada individuo es seleccionado de forma independiente e idéntica a como lo son los demás.

Los valores posibles de cada variable aleatorio junto con las probabilidades conlos que los toma, se denomina distribución o les de probabilidad de la variable aleatoria en estudio, o más brevemente, **Modelo probabilístico**. Habitualmente se supondrá un modelo probabilístico que se crea es un reflejo lo máas exactoposible del fenómeno aleatorio que se este estudiando.

De este modo, identificada la variable aleatorio y supuesto una ley de probabilidad o modelo probabilístico que sija dicha variable, lo que queda es hacer inferencia sobre los parámetrosdesconocidos que definene el comportamiento del modelo probabilístico.

## Conceptos fundamentales de la estadística descriptiva

Cada individuo de la población posee caracteres o caracteristicas que lo describen. Estos caracteres pueden ser cuantitativos o cualitativos. Las caracteristicas observadas tienen asociados datos que denominaremos modalidades, por eejmplo hombre y mujer en la variable sexo en species co dimorfismo sexual. Estas características deben ser sijuntas y colectivament exhaustivas.

Nótese que convencionalmente las caracteristicas se refieren como variables, mientras que los valores de la variable serían las modalidades.

En ocasiones podemos considerar organizar los datos por intervalos o clases:
$$
[c_{0} - c_{1}]
$$
E; centor o marca de clase corresponde al interbalo ubicaden punto medio del intervalo.

Para definir la amplitud del intervalo se sugiere utilizar el criterio de Sturges (1926) para definir el ñumero k e intervalos
$$
k = 1 + 3.322 \ log_{10^n}
$$
### Distribuciones unidimensionales de frecuencias

Se llama frecuencia total al número de datos n. Frecuencua absoluta de ni de la modalidad Mi al número de datos que representa la modalidad Mi

Si existen k modalidades posibles, se verificará que:

$$
\sum_{i=1}^{k} n_{i} = n_{1} + \dots + n_{k} = n
$$
La frecuencia realtiva fi de la modalidad Mi, o del intervalo Ii, de X es el coeficiente

$$
f_I = \frac{n_i}{n}
$$

$$
\sum_{i=1}^k f_i = f_1 + \dots + f_k = 1
$$
La frecuencia absoluta acomulada Ni hasta la modalidad Mi, al igual que la frecuencia relativa, se definene
$$
N_i = n_1 + \dots + ni = \sum_{j=1}^i n_j
$$
$$
F_i = f_1 + \dots + fi = \sum_{j=1}^i f_j
$$

#### Ejemplo

Tomemos una variable continua definida como x1

```r
x1 <- c(10.6,12.5,11.1,9.2,11.5,9.9,11.9,11.6,14.9,12.5,12.5,12.3,
12.2,10.8,16.5,15,10.3,12.4,9.1,7.8,11.3,12.3,9.7,12,11.8,12.7,
11.4,9.3, 8.6, 8.5, 10.1, 12.4, 11.1, 10.2)

table(x1)
```

La tabla de frecuencia sin intervalos puede no brindarnos mucha información clara.

```r
# 1. Calcular el número de intervalos según Sturges
k <- nclass.Sturges(x1)

# 2. Agrupar los datos usando el valor numérico de k
intervalos <- cut(x1, breaks = k, include.lowest = TRUE)

# 3. Generar la tabla de frecuencias
frec_abs <- table(intervalos)
frec_rel <- prop.table(frec_abs)

tabla_frecuencia <- data.frame(
  Intervalo = names(frec_abs),
  ni = as.numeric(frec_abs),
  fi = round(as.numeric(frec_rel), 4),
  Ni = cumsum(as.numeric(frec_abs)),
  Fi = round(cumsum(as.numeric(frec_rel)), 4)
)

print(tabla_frecuencia)
```

### Medidas

Centralidad: Media, mediana, moda, cuantiles
Dispersión: Recorrdio, varianza, desivaciön típica, coeficiente de variación de Pearson.
Medidas de asimetría: Una distribución es simétrica cunado su mediana, moda y media aritmetica coincidan.
Es asimetrica a la derecha si las frecuencias descienden mas lentamente por la derecha que por la izquierda.Sidescienden mas lentamente por la izquierda hay asimetría a la izquierda.

Coeficiente de asimetría de Pearson

$$
A_p  = \frac{a-M_d}{s}
$$
Siendo a la media, Md la moda, s la desviación típica. El coeficiene es cero cuando la distribuciön es simétrica, positivo cuando existe asimetría a la derecha y negativo cunado existe asimetría a la izquierda.

Coeficiente de asimetría de Fisher

$$
A_f = \frac{\sum_{i=1}^k (x_i - a)^3 }{n S^3}
$$
siendo xi los valores de la variable, a la media, S la cuasidesviación típica. La interpretración del coeficiente de Fisher es análoga a la de Pearson.

#### Ajuste por mínmos cuadrados

Definimos una representación lineal de la relación entre dos variables X e Y

$$
y_{t_{i}} = \beta_{0} + \beta_{1}x_{i}
$$
Donde yti es una aproximación estimada del valor empírico de yt.

Debemos entonces determinar los valores de los parámetros beta que minimicen la suma de los cuadrados de las desviaciones ei entre los valores observados yi y los teóricosdados por la recta yti

$$
\sum_{i=1}^n e_{i}^2 = \sum_{i=1}^n (y_{i}-y_{t_{i}})^2 = \sum_{i=1}^n (y_{i} - \beta_{0} - \beta_{1} x_{1}) ^2
$$
este problema se resuelve considerando la ecuación anterior como una funcion de beta 0, derivando respoecnto a bea 0 e igualando a cero dicha ecuación, lo mismo con beta 1. Esto nos da un sistema de ecucaciones con dos incognitas de donde despejamos y obtenemos los valores para beta 1 y 0.

```r
df = iris
ajus <- lm(df$Sepal.Length ~ df$Petal.Length)
ajus

plot(df$Petal.Length, df$Sepal.Length, pch = 16,
     xlab = "Petal Lenght", ylab = "Sepal Length")

abline(ajus, lty=2,col=4, lwd = 2)
```

Si las variables no se relacionan linealmente, podemos probar otras configuraciones. Por ejemplo sea mejor un ajuste de un modlo potencial
$$
y_{ti} = ab^{x_{i}}
$$
con b <1. Es decir, se ajustase mejor a los datos una recta de la forma
$$
ln \ y_{ti} = ln \ a + b \ ln \ x_{i}
$$

```r
# 1. Ajustar el modelo potencial
ajus_pot <- lm(log(Petal.Length) ~ log(Sepal.Length), data = df)

A <- exp(coef(ajus_pot)[1])
B <- coef(ajus_pot)[2]

# 2. Graficar datos reales
plot(df$Sepal.Length, df$Petal.Length,
     pch = 16, xlab = "Sepal Length", ylab = "Petal Length",
     main = "Modelo Potencial: Y = A * X^B")

# 3. Dibujar la curva y = A * x^B
x_grid <- seq(min(df$Sepal.Length), max(df$Sepal.Length), length.out = 100)
y_pred <- A * (x_grid ^ B)

lines(x_grid, y_pred, col = "purple", lwd = 2, lty = 2)
```

O tambien un modelo exponencial

```r
# 1. Ajuste del modelo lineal en escala logarítmica: ln(Y) ~ X
ajus_exp <- lm(log(Petal.Length) ~ Sepal.Length, data = df)
summary(ajus_exp)

# Extraer parámetros ajustados
a <- coef(ajus_exp)[1] # Intercepto lineal: ln(A)
b <- coef(ajus_exp)[2] # Pendiente: B
A <- exp(a)             # Escala original: A = e^(ln(A))

# 2. Gráfico en escala original
plot(df$Sepal.Length, df$Petal.Length,
     pch = 16,
     xlab = "Sepal Length",
     ylab = "Petal Length",
     main = "Ajuste Exponencial: Petal.Length vs Sepal.Length")

# 3. Superponer la curva exponencial
x_grid <- seq(min(df$Sepal.Length), max(df$Sepal.Length), length.out = 100)
y_pred <- A * exp(b * x_grid) # O equivalente: exp(predict(ajus_exp, newdata = data.frame(Sepal.Length = x_grid)))

lines(x_grid, y_pred, col = "red", lwd = 2, lty = 2)
```
