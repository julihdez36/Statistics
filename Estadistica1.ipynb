# Introducción a R

R es un software clónico del paquete (no gratuito) S-plus, que es un compendio de aplicaciones estadísticas que utilizan el lenguaje S, diseñado por *Bell Laboratories*.

El elemento básicos de R son los objetos: ficheros capaces de ser editaso y ejecutados. Los objetos se crean a través del operador de asignación:


```r
a <- 1
a+a
```

Los dos tipos de objetos más utilizados son el dato y la función. Estas últimas estan seguidas de paréntesis que incluyen argumentos. Si sólo se ejecuta el nombre se obtendrá la definición y si se ejecuta con un "?" antes, se puede acceder a la documentación.

Es posible afirmar que R es un lenguaje funcional, en el sentido de que sus programas se presentan como funciones escritas en su lenguaje.

La función *object* nos permite listar los objetos R existentes

```r
b <- "hola"
c <- FALSE
objects()
```

Pos su parte *rm* nos permite eliminar objetos

```r
rm(a)
objects()
```

La estructura básica de las funciones de R se presentan a continuación:


```r
f <- function(x){2*x}
f(12)
```

## Datos en R

En R un dato es el resultado de ejecutar una expresión R, es decir, un tipo de objeto. Cada dato tiene asociados determinados atributos, el mas importante es su modo: lógico, numérico, complejo, caracter. A su vez, hay diferentes tipos de datos: vector, matriz, factor, data frame, lista.

### Tipos de datos:

#### Vectores:

El vector es el tipo de dato mas utilizado en R, especialmente como argumento de funciones.

Todos los elementos de un vector deben ser del mismo modo y su cantidad constituye la longitud:

```r
x <- c(1:6)
mode(x)
length(x)
```

#### Factores

Es un vector de datos no numéricos formado por datos procedentes de categorías, como la clasificación por sexo.

#### Matrices

Disposición bidimensional en donde, al igual que con los vectores, todos los elementos deben ser del midmo modo.

```r
A = matrix(c(1:16),ncol = 4,nrow = 4, byrow = F)
A
```

También podemos formar matrices combinando vectores. Además del modo y la longitud, otro atributo importante en matrices es la dimensión

```r
x <- 1:5
y <- 6:10
B = rbind(x,y) # cbind
B
length(B)
dim(B)
```


podemos poner nombres a los ejes de la matriz


```r
C <- matrix(1:8, ncol = 4, dimnames = list(c("Indiv. 1","Indiv. 2"),c("car.1","car.2","car.3","car.4")))
C
```


#### Data frame

Esta estructura de datos permite datos de diferente modo, por lo que es una de las mas usadas en el analisis de datos.

#### Listas

El objeto-dato de R mas felxible es la lista, el cual puede admitir datos de diferentes modos y de diferentes longitudes, por lo que son habitualmente el resultado final en donde se van incorporando los otros tipos de datos.

La mayoría de de funciones de R que realizan analisis estdítico presentan sus resultados en una lista: datos originales, valores ajustados, residuos, estadísticos de contraste, p-valor e inclusive el método empleado.


### Gráficos:

Las funciones gráficas se dividen en funciones de alto nivel y bajo nivel. Con las primeras se obtiene un gráfico nuevo, con las segundas podemos hacer modificaciones de un gráfico ya existente.

```r
x <- seq(-2*pi, pi, len = 100) # vector de abscisas
plot(x, sin(x), type = "l", main = "Función Seno")
abline(1,0)
```
